```python
import pandas as pd
from tabulate import tabulate
```

```python
df = pd.read_excel('sku_tien_xu_ly_lan1 (1).xlsx')
df_label = pd.read_csv('sku_cluster_labels.csv')

char = df[['ABC', 'FSN']].copy()
char['ABC'] = char['ABC'].replace({'Thiếu giá' : 'U'})
char['FSN_ABC'] = char['FSN'] + char['ABC']
char['cluster'] = df_label['cluster']
cleaned_char = char[char['cluster'] != -1].copy()
```

```python
print("Tổng số SKU trước khi lọc nhiễu: ",  len(char))
print("Tổng số SKU sau khi lọc nhiễu: ",    len(cleaned_char))
```

```python
p_label = cleaned_char['FSN_ABC'].value_counts(normalize = True)
p_label.index.name = 'FSN_ABC'

p_label_given_cluster = (cleaned_char.groupby('cluster')['FSN_ABC'].value_counts(normalize = True))
p_label_given_cluster.index.names = ['cluster', 'FSN_ABC']

lift = p_label_given_cluster.div(p_label, level = 'FSN_ABC')

lift_matrix = lift.unstack('FSN_ABC', fill_value = 0)
lift_matrix = lift_matrix.sort_index()

print(tabulate(lift_matrix, headers = 'keys', tablefmt = 'grid'))
lift_matrix.to_excel('Lift_Matrix.xlsx')
```
